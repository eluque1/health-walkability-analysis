## Topic 1: Feature Importance
[Hyperlink](
https://colab.research.google.com/drive/1F8mymHyM6xrqbNXzH0Tst-22sntcRChE#scrollTo=aEvJR57QifF1&line=12&uniqifier=1)
### Why we used this concept
Our dataset contains dozens of predictors from two sources: CDC PLACES (health indicators) and the EPA Walkability Index (built-environment features). Simply training a model does not show which predictors actually drive walkability.
Feature importance allows us to separate strong, weak, and irrelevant predictors and to compare linear vs. nonlinear relationships.

### How we implemented it
We computed feature importances using two different models:

1. **Ridge Regression (baseline)**  
   - Extracted coefficients from `ridge.best_estimator_.coef_`

2. **Random Forest (model 2)**  
   - Used `best_rf.feature_importances_`

3. **XGBoost (model 3)**
   - Extracted importance values from `xgb_model.best_estimator_.feature_importances_`

4. For all 3 models:
   - Sorted by asolute magnitude
   - Visualized the top 15 coefficients using a bar chart

### Results & Interpretation
Across all three models, the K-Means health clusters (HealthCluster_0–HealthCluster_3) were consistently the strongest predictors of walkability. This indicates that a county’s overall health profile (rather than any single condition) is most strongly associated with how walkable it is. Across all three models, `HealthCluster_3` and `HealthCluster_2` consistently rank as the strongest predictors of walkability, which means counties with healthier populations tend to have higher walkability scores. The consistency of the top predictors across Ridge, Random Forest, and XGBoost confirms that the relationship between health and walkability is stable across modeling approaches, strengthening the reliability of our findings.

## Topic 2: Ensemble Models (RandomForest + XGBoost)
[RandomForest](https://colab.research.google.com/drive/1F8mymHyM6xrqbNXzH0Tst-22sntcRChE#scrollTo=5v__6qTrGzG3&line=3&uniqifier=1)

[XGBoost](https://colab.research.google.com/drive/1F8mymHyM6xrqbNXzH0Tst-22sntcRChE#scrollTo=nQIieSujG25g&line=13&uniqifier=1)

### Why we used this concept
Walkability is shaped by complex, nonlinear interactions among population density, land-use mix, intersection structure, and access to transit. Linear models cannot capture these patterns. Ensemble models such as RandomForest and XGBoost allow us to model nonlinear relationships, feature interactions, and heterogeneous effects across counties, and improve predictive performance relative to the baseline.
### How we implemented it
Random Forest (Model 2)
- Tuned parameters (`n_estimators`, `max_depth`, `min_samples_split`, `min_samples_leaf`) using GridSearchCV
- Trained the best model on the full training set
- Evaluated with R², MSE, MAE
- Extracted feature importances for interpretation

XGBoost (Model 3)
- Tuned 48 hyperparameter combinations (`max_depth`, `learning_rate`, `subsample`, `colsample_bytree`, `n_estimators`)
- Selected best model based on R²
- Evaluated with R², MSE, MAE
- Compared performance directly to RandomForest and Ridge

### Results & Interpretation
Random Forest performed similarly to Ridge, showing small differences in error metrics and relying heavily on the top health clusters. XGBoost achieved the best (but still modest) improvement, with slightly higher R² and lower MAE. The overall similarity across models suggests that health variables provide moderate predictive power, with boosting offering only incremental gains.

### Why we used this concept
All three of our models depend on hyperparameters that strongly influence performance. Using defaults leads to underfitting or overfitting and makes comparisons unfair. Hyperparameter tuning ensures each model is evaluated at its best and increases the reliability of our conclusions.


### How we implemented it
We used `GridSearch` with cross-validation for all three models:

- Ridge (baseline)
  - Tuned alpha across [0.01, 0.1, 1, 10, 100]
  - Selected the regularization level w/ best validation performance
- Random Forest (model 2)
  - Tuned depth, number of trees, and splitting rules
  - Used a subset for speed, retrained final model on full training data
- XGBoost (model 3)
  - Tuned 48 combinations of tree depth, learning rate, subsampling, and column sampling
  - Selected best parameters based on R² score

### Results & Interpretation
Hyperparameter tuning allowed each model to run with reasonable, well-chosen settings, making the performance comparison fair. With tuned parameters, Ridge, Random Forest, and XGBoost all achieved similar results (R² around 0.41–0.43 with comparable MSE/MAE).
